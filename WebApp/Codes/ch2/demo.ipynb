# Chapter 2 Engineering Economy Factors

This notebook demonstrates the `factors.py` library which contains standard mathematical functions, amortization schedules, and interactive plotting.

```python
import factors
import pandas as pd

# Example 1: Bart's Car Loan
principal = 120000
rate = 0.07 / 12  # 7% annual, compounded monthly
periods = 60

payment = factors.calculate_ap(rate, periods, p=principal)
print(f"Monthly Payment: ${payment:,.2f}")
```

## Generating the Amortization Schedule
Because we return a pandas DataFrame, the schedule renders beautifully as an HTML table natively.

```python
df_schedule = factors.generate_amortization_schedule(principal, rate, periods)
display(df_schedule)
```

## Interactive Plotting
Using Plotly, we can visualize the amortization breakdown or general cash flows.

```python
fig = factors.plot_amortization_breakdown(df_schedule)
fig.show()
```

## Composite Gradients
Real-world gradients usually have a uniform base amount plus a strictly increasing gradient amount. We can calculate the total composite value easily.

```python
# Example: Cobot Maintenance
base_amount = 1000
gradient_amount = 200
rate = 0.08
periods = 5

p_total = factors.calculate_composite_pg(rate, periods, a=base_amount, g=gradient_amount)
a_total = factors.calculate_composite_ag(rate, periods, a=base_amount, g=gradient_amount)

print(f"Total Present Worth: ${p_total:,.2f}")
print(f"Equivalent Uniform Annual Cost: ${a_total:,.2f}")
```
